```python
aws_response = {
    "Reservations": [
        {
            "Instances": [
                {
                    "InstanceId": "i-123",
                    "InstanceType": "t3.micro",
                    "State": {"Name": "running"}
                }
            ]
        }
    ]
}
# access the instance id

# my approach

for dict_item in aws_response.values():
    print(dict_item)
    for l1_item in dict_item:
        print(l1_item)
        print(type(l1_item))
        for l2_item in l1_item.values():
            print(l2_item)
            print(type(l2_item))
            for l3_item in l2_item:
                print(l3_item)
                print(type(l3_item))
                for key,value in l3_item.items():
                    if 'InstanceId' in key:
                        print(value)

# industry approach
for reservation in aws_response["Reservations"]:
    for instance in reservation["Instances"]:
        print(instance["InstanceId"])
```

```python
# extract instance state
aws_response = {
    "Reservations": [
        {
            "Instances": [
                {
                    "InstanceId": "i-123",
                    "InstanceType": "t3.micro",
                    "State": {"Name": "running"}
                }
            ]
        }
    ]
}

# Your code here
for reservation in aws_response['Reservations']:
    print(reservation)
    for instance in reservation['Instances']:
        print(instance)
        print(instance["State"])
        for key,value in instance["State"].items():
            if 'running' in value:
                print(value)
        # print(instance['State'] if 'running' in instance["State"]['Name'] else 0)
        
# pythonic way
for reservation in aws_response['Reservations']:
    for instance in reservation['Instances']:
        if instance["State"]["Name"] == "running":
            print(f"Instance {instance['InstanceId']} is running")
```

```python
data = {
    "Instances": [
        {"InstanceId": "i-1", "State": "running"},
        {"InstanceId": "i-2", "State": "stopped"},
        {"InstanceId": "i-3", "State": "running"}
    ]
}

# Your code here
for instance in data["Instances"]:
    print(f"{instance['InstanceId']} {instance['State']}")
# expected output
''' 
i-1 running
i-2 stopped
i-3 running
'''

# Filter Only Running Instances
for instance in data["Instances"]:
    if instance["State"] == 'running':
        print(instance["InstanceId"])
```
