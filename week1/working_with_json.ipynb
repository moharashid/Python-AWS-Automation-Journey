```python
import json
```

```python
data = {
    "Instances": [
        {"InstanceId": "i-1", "State": "running"},
        {"InstanceId": "i-2", "State": "stopped"}
    ]
}

json_data = json.dumps(data)

print(json_data)
print(type(json_data))

# with indent
json_data_indented = json.dumps(data,indent=4)
print(json_data_indented)
```

```python
# saving the json to a file
data = {
    "Instances": [
        {"InstanceId": "i-1", "State": "running"},
        {"InstanceId": "i-2", "State": "stopped"},
        {"InstanceId": "i-3", "State": "running"}
    ]
}

with open("ec2_json.json", "w") as file:
    json.dump(data, file, indent=5)
    print('JSON copied to file successfully!')
```

```python
with open("ec2_json.json", "r") as file:
    data_2 = json.load(file)

print(json.dumps(data, indent=4))
```

```python
with open("ec2_json.json", "r") as file:
    data_2 = json.load(file)

# print(data)
for instance in data_2["Instances"]:
    print(f'Instance {instance["InstanceId"]} is {instance["State"]}')
    
# print onlyrunning instance
for instance in data_2["Instances"]:
    if instance["State"] == "running":
        print(f'Running instance: {instance["InstanceId"]}')
```
